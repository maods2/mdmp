# MDMP quickstart

This notebook shows a minimal end-to-end example of fitting an MDM model on a small synthetic dataset and visualizing the learned DAG structure.

```python
import sys

sys.path.append("..")
import pandas as pd

from mdmp import MDM, plot_arcs, plot_dag, list_datasets, load_dataset
```

```python
list_datasets()
```

```python
data = load_dataset("covid_regional_timeseries")

# Fit MDM model
model = MDM(data, method="hc", verbose=True)
```

```python
# Plot the learned DAG
plot_dag(model, plot_type="graph", layout_seed=5);
```

```python
plot_arcs(model, plot_type="connections");
```
