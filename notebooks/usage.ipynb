# MDMP quickstart

This notebook shows a minimal end-to-end example of fitting an MDM model on a small synthetic dataset and visualizing the learned DAG structure.

```python
import sys
sys.path.append("..")
import numpy as np
import pandas as pd

from mdmp import MDM, plot_dag, plot_arcs, plot_marginal, plot_stream, plot_idag
```

```python
data = pd.read_csv("simulated_dag_time_series.csv")

# Fit MDM model
model = MDM(data, method="hc", verbose=True)
```

```python
# Plot the learned DAG
plot_dag(model, plot_type="graph");
```

```python
plot_arcs(model, plot_type="connections");
```
