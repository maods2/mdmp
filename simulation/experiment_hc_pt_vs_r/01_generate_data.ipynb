# 01 — Generate simulated data

Generate **300 replications per DAG** (3-variable and 5-variable) using the
existing `simulate_dags.run_simulations` helper.

**Fixed scenario:** `W=0.01`, `V=100.0`, `T=200`  
**Seed policy:** `base_seed=1564`; individual `i` uses seed `1564 + i`.

Data is written to `data/` and shared with both Python and R notebooks.

```python
import glob
import os
import sys

sys.path.insert(0, os.path.abspath(".."))
from simulate_dags import run_simulations

W = 0.01
V = 100.0
T = 200
BASE_SEED = 1564
N_REPLICATIONS = 300

run_simulations(
    output_dir="data/",
    base_seed=BASE_SEED,
    n_individuals=N_REPLICATIONS,
    w_values=(W,),
    v_values=(V,),
    t_values=(T,),
)
```

```python
data_dir = "data/"
prefix_3 = f"dag_3var_W{W}_V{V}_T{T}"
prefix_5 = f"dag_5var_W{W}_V{V}_T{T}"

files_3var = glob.glob(os.path.join(data_dir, f"{prefix_3}_ind*.csv"))
files_5var = glob.glob(os.path.join(data_dir, f"{prefix_5}_ind*.csv"))

true_adj_3 = os.path.join(data_dir, f"{prefix_3}_true_adjacency.csv")
true_adj_5 = os.path.join(data_dir, f"{prefix_5}_true_adjacency.csv")

assert len(files_3var) == N_REPLICATIONS, f"Expected {N_REPLICATIONS} 3var files, got {len(files_3var)}"
assert len(files_5var) == N_REPLICATIONS, f"Expected {N_REPLICATIONS} 5var files, got {len(files_5var)}"
assert os.path.isfile(true_adj_3), f"Missing {true_adj_3}"
assert os.path.isfile(true_adj_5), f"Missing {true_adj_5}"

print(f"3var datasets: {len(files_3var)}")
print(f"5var datasets: {len(files_5var)}")
print(f"True adjacency files: OK")
```
